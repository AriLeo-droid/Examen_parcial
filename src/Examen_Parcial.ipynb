```python
import pandas as pd 
import numpy as np 
import missingno as msno
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/sensores_industriales.csv", index_col="id_registro")
df.reset_index(drop=True, inplace=True)
df.index += 1
```

```python
df.head()
```

# 1. Exploración inicial

```python
# Cantidad de registros y dimensiones en general

#Columnas (Variables):
print(df.shape[1], "variables contadas en el dataset")
print()

#Cantidad de filas (observaciones)
print(df.shape[0], "observaciones contadas en el dataset")
print()

#Dimensión general:
print(df.shape[0]*df.shape[1])
print()

#Cantidad de sensores distintos:
print("Sensores diferentes en la planta:")
sensores = df.value_counts("id_sensor")
print(sensores)
```

# 2. Cálculos iniciales

```python
# Temperatura promedio de cada planta:
for sensor, num in df.groupby("id_sensor"):
    promedio = num["temperatura_c"].mean()
    print(f"Sensor: {sensor}")
    print(f"promedio de temperatura: {promedio}")
    print()
```

# 3. Máximo e información

```python
max = df.loc[df["temperatura_c"].idxmax()]
print("registro de máxima temperatura registrada:")
print(max)
```

# 4. Rangos térmicos

```python
temp_fil = df[df["temperatura_c"] > 85]

for indice, fila in temp_fil.iterrows():

    print(
        f"registro: {indice} | "
        f"sensor: {fila['id_sensor']} | "
        f"temperatura: {fila['temperatura_c']}"
    )
```

# 5. Identificar la planta con más alertas de temperatura

```python
cont_temp = temp_fil["id_sensor"].value_counts()
 
sensor = cont_temp.idxmax()
cantidad = cont_temp.max()
 
print(f"El sensor: {sensor} tiene mayor cantidad de alertas de temperatura, con {cantidad} alertas")
```

# 6. Exportación de alertas

```python
temp_fil.to_csv("../Export/Alertas_térmicas_sensores_industriales.csv", encoding="utf-8")
```

# 7. Empatados

```python
emp = temp_fil[temp_fil["temperatura_c"].duplicated(keep=False)]

for temp, sensor in emp.groupby("temperatura_c"):

    print(f"\n Temperatura duplicada: {temp}")

    for indice, fila in sensor.iterrows():

        print(
            f"registro: {indice} | "
            f"id_sensor: {fila['id_sensor']} | "
            f"temperatura: {fila['temperatura_c']}"
        )
```
